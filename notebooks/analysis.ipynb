# Data profiling
Inspect columns, missing values, duplicates, rating balance and review length.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/hotel_reviews.csv")
display(df.head())
print(df.info())
print("Missing values:\n", df.isna().sum())
print("Duplicate reviews:", df["review"].duplicated().sum())
df["word_count"] = df["review"].fillna("").str.split().str.len()
display(df["word_count"].describe())
df["rating"].value_counts().sort_index().plot.bar(title="Rating distribution")
plt.show()
```
